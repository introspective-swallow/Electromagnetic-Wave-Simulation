```python
import numpy as np
import matplotlib.pyplot as plt

FIGS = ""
```

```python

c = 1 # speed of light

def xi(N, S):
    xi_value = 1 + (1/S)**2 * (np.cos(2*np.pi*S/N)-1)
    #if xi_value >= -1:
        #print("Warning: xi is greater than -1", xi_value)
    return xi_value

def numerical_phase_velocity(N, S):
    N_transition = 2*np.pi*S / np.arccos(1-2*S**2)
    if N>N_transition:
        return 2*np.pi / (N * np.arccos(xi(N, S))) * c
    elif N< N_transition:
        return (2/N) * c

def amplitude_multiplier(N, S):
    N_transition = 2*np.pi*S / np.arccos(1-2*S**2)
    if N>N_transition:
        return 1
    elif N< N_transition:
        xi_value = xi(N, S)
        return - xi_value - np.sqrt(xi_value**2 - 1)

# Plot
S = 0.7
print("Transition point: ", 2*np.pi*S / np.arccos(1-2*S**2))
N_linspace = np.linspace(1, 10, 10000)
v = [numerical_phase_velocity(N, S) for N in N_linspace]

plt.plot(N_linspace, v)

# Plot the speed of light and write the value
plt.hlines(c, 0, 10, color="black", linestyle="--")
plt.text(0.5, c*1.05, r"$c$")

plt.xlabel(r"$N_{\lambda}$")
plt.ylabel(r"$\overline{v}_p$")
plt.savefig(FIGS + "numerical_phase_velocity.png")
plt.show()

plt.figure()
N2_linspace = np.linspace(1, 10, 1000)
at = [amplitude_multiplier(N, S) for N in N2_linspace]
plt.plot(N2_linspace, at)
plt.xlabel(r"$N_{\lambda}$")
plt.ylabel(r"$e^{\overline{k}_{imag}\Delta x}$")
plt.savefig(FIGS + "amplitude_multiplier.png")
plt.show()
```

```python
S = 1/np.sqrt(2)
N_transition = 2*np.pi*S / np.arccos(1-2*S**2)
print("Transition point: ", N_transition)
# Plot xi
N_linspace = np.linspace(1, 1000, 10000)
xi_values = [xi(N, S) for N in N_linspace]
plt.plot(N_linspace, xi_values)
```

```python
def numerical_phase_velocity_improved(N, S):
    xi_value = xi(N, S)
    if xi_value >= -1:
        return 2*np.pi / (N * np.arccos(xi(N, S))) * c
    elif xi_value < -1:
        return (2/N) * c

def amplitude_multiplier_improved(N, S):
    xi_value = xi(N, S)
    if xi_value >= -1:
        return 1
    elif xi_value < -1:
        return - xi_value - np.sqrt(xi_value**2 - 1)
S = 1

print("Transition point: ", 2*np.pi*S / np.arccos(1-2*S**2))
N_linspace = np.linspace(0, 1, 10000)
v = [numerical_phase_velocity(1/N, S) for N in N_linspace]

plt.plot(N_linspace, v)

# Plot the speed of light and write the value
plt.hlines(c, 0, 1, color="black", linestyle="--")
plt.text(0.3, c*1.05, r"$c$")

plt.xlabel(r"$1/N_{\lambda}$")
plt.ylabel(r"$\overline{v}_p$")
#plt.savefig(FIGS + "numerical_phase_velocity.png")
plt.show()

plt.figure()

N3_linspace = np.linspace(0,1, 1000)
at = [amplitude_multiplier_improved(1/N, S) for N in N3_linspace]
plt.plot(N3_linspace, at)
plt.xlabel(r"$1/N_{\lambda}$")
plt.ylabel(r"$e^{\overline{k}_{imag}\Delta x}$")
#plt.savefig(FIGS + "amplitude_multiplier.png")
plt.show()
```
